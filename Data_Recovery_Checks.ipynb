```python

import os

home = os.path.expanduser("~")
print("Your Windows user folder:", home)

matches = []

for root, dirs, files in os.walk(home):
    # Skip folders that can contain lots of irrelevant files
    dirs[:] = [d for d in dirs if d not in [
        "AppData", "node_modules", ".conda"
    ]]

    for file in files:
        if file.lower().endswith(".zip"):
            matches.append(os.path.join(root, file))

print("ZIP files found:")
for path in matches:
    print(path)

if not matches:
    print("No ZIP files found in this user folder.")
```

```python

import pandas as pd
import sqlite3
import os
import time

csv_path = r"C:\Users\Medion\original_data_from_zip\data\purchases.csv"
db_path = r"C:\Users\Medion\inventory.db"

start_time = time.time()

# Read the original CSV in chunks to reduce memory usage
first_chunk = True

with sqlite3.connect(db_path) as conn:
    for chunk in pd.read_csv(csv_path, chunksize=100_000):
        chunk.to_sql(
            "purchase",
            conn,
            if_exists="replace" if first_chunk else "append",
            index=False
        )
        first_chunk = False
        print(f"Loaded {len(chunk):,} rows")

    count = pd.read_sql_query(
        "SELECT COUNT(*) AS total_rows FROM purchase",
        conn
    ).iloc[0, 0]

print(f"\nTotal rows in purchase table: {count:,}")
print(f"Time taken: {(time.time() - start_time) / 60:.2f} minutes")
```
