```python
def ingest_db(df, table_name, engine):
    df.to_sql(
        table_name,
        con=engine,
        if_exists='replace',
        index=False
    )
```

```python

def load_raw_data():
    for file in os.listdir():
        if file.lower().endswith('.csv'):
            try:
                df = pd.read_csv(file)

                logging.info(
                    f"Loaded {file} with shape {df.shape}"
                )

                ingest_db(df, file[:-4], engine)

                logging.info(f"Inserted {file} into database")
                print("Loaded and inserted:", file, df.shape)

            except Exception:
                logging.exception(f"Failed to process {file}")
                print("Error processing:", file)
```

```python
import time
```

```python
start_time = time.time()

load_raw_data()

end_time = time.time()

total_time = (end_time - start_time) / 60

print(f"Total loading time: {total_time:.2f} minutes")
```
